# 01. Dataset Validation

This notebook verifies the integrity of the raw Udemy courses dataset file (`ml-training/datasets/raw/udemy_courses.csv`).

```
import os
import pandas as pd

dataset_path = os.path.join("..", "..", "datasets", "raw", "udemy_courses.csv")

if not os.path.exists(dataset_path):
    raise FileNotFoundError(f"Dataset not found at {dataset_path}. Please download the Kaggle Udemy Courses dataset and place it in the datasets/raw folder.")

df = pd.read_csv(dataset_path)
print(f"Total rows: {len(df)}")
print(f"Columns: {list(df.columns)}")
print(f"Null counts:\n{df.isnull().sum()}")
print(f"Duplicate rows: {df.duplicated().sum()}")
print(f"Unique subjects/categories: {df['subject'].nunique()}")
print(df['subject'].value_counts())
```
