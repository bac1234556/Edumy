# 04. Train Linear SVM

Trains the Linear SVM baseline classifier, applies parameter searches over C, and logs performance.

```
import os
import pandas as pd
import pickle
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import LinearSVC
from sklearn.metrics import classification_report

processed_path = os.path.join("..", "..", "datasets", "processed", "course_classification_processed.csv")
df = pd.read_csv(processed_path)

train_df = df[df['split'] == 'train']
val_df = df[df['split'] == 'val']

# TF-IDF Fit strictly on Train set
tfidf = TfidfVectorizer(max_features=5000, ngram_range=(1, 2), min_df=2, max_df=0.95)
X_train = tfidf.fit_transform(train_df['cleaned_text'])
X_val = tfidf.transform(val_df['cleaned_text'])
y_train = train_df['encoded_label']
y_val = val_df['encoded_label']

# Evaluate parameter tuning over C
for c_val in [0.1, 0.5, 1.0, 2.0]:
    svm = LinearSVC(C=c_val, random_state=42)
    svm.fit(X_train, y_train)
    val_acc = svm.score(X_val, y_val)
    print(f"C={c_val} -> Validation Accuracy: {val_acc:.4f}")
```
