# 01. Dataset Validation

This notebook verifies that all OULAD raw tables are loaded correctly, prints file statistics, and checks the schema.

```python
import os
import pandas as pd

raw_dir = os.path.join("..", "..", "datasets", "raw")
required_files = [
    "courses.csv",
    "assessments.csv",
    "studentAssessment.csv",
    "studentInfo.csv",
    "studentRegistration.csv",
    "studentVle.csv",
    "vle.csv"
]

print("Checking raw dataset files...")
for f in required_files:
    path = os.path.join(raw_dir, f)
    if not os.path.exists(path):
        raise FileNotFoundError(f"Missing required file: {f} at {path}")
    size_mb = os.path.getsize(path) / (1024 * 1024)
    print(f"- {f}: Found, Size: {size_mb:.2f} MB")

# Read and print shapes/schemas
df_courses = pd.read_csv(os.path.join(raw_dir, "courses.csv"))
df_reg = pd.read_csv(os.path.join(raw_dir, "studentRegistration.csv"))
df_vle = pd.read_csv(os.path.join(raw_dir, "studentVle.csv"), nrows=1000) # read sample to check

print(f"\nCourses shape: {df_courses.shape}")
print(f"Student Registration shape: {df_reg.shape}")
print(f"Student VLE (sample) shape: {df_vle.shape}")

print("\nDataset validation PASS.")
```
