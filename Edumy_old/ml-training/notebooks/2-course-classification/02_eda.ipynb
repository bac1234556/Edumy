# 02. Exploratory Data Analysis (EDA)

This notebook computes length distributions, category balances, and renders charts.

```
import os
import pandas as pd
import matplotlib.pyplot as plt

dataset_path = os.path.join("..", "..", "datasets", "raw", "udemy_courses.csv")
df = pd.read_csv(dataset_path)

# 1. Category Distribution
cat_counts = df['subject'].value_counts()
plt.figure(figsize=(8, 5))
cat_counts.plot(kind='bar', color='skyblue')
plt.title('Category Distribution')
plt.ylabel('Count')
plt.xlabel('Subject')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 2. Title Lengths
df['title_len'] = df['course_title'].apply(lambda x: len(str(x)))
df['word_count'] = df['course_title'].apply(lambda x: len(str(x).split()))

plt.figure(figsize=(10, 4))
plt.hist(df['word_count'], bins=20, color='salmon', edgecolor='black')
plt.title('Course Title Word Count Distribution')
plt.xlabel('Word Count')
plt.ylabel('Frequency')
plt.show()

print("Word count statistics:")
print(df['word_count'].describe())
```
