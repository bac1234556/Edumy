# 02. Exploratory Data Analysis (EDA)

Analyzes sparsity densities and user-item clicks distributions.

```python
import os
import pandas as pd

registration_path = os.path.join("..", "..", "datasets", "raw", "studentRegistration.csv")
df_reg = pd.read_csv(registration_path)

u_count = df_reg['id_student'].nunique()
i_count = df_reg['code_module'].nunique()
n_count = len(df_reg)
sparsity = 1.0 - (n_count / (u_count * i_count))

print(f"Users (Unique Students): {u_count}")
print(f"Items (Unique Courses): {i_count}")
print(f"Interactions (Registrations): {n_count}")
print(f"Sparsity: {sparsity*100:.2f}%")
print(f"Density: {(1 - sparsity)*100:.2f}%")
```
